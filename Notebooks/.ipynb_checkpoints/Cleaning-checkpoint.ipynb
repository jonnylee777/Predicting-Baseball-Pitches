# Pitch Prediction Project

In this project, we will attempt to predict what pitch a given pitcher will throw in a given situation using historical data. We seek to use various classification models to see how accurately we can predict a pitcher's next pitch given the situation using historical data. Different models to consider include: Logistic Regression, Random Forest, Gradient Boosting, and Support Vector Machines.
In our first iteration, we choose Kevin Gausman as our pitcher case study. He has three pitches that he used in 2025: 4-seam fastball (52.1%), splitter (38.8%), and slider (9.1%).

We found a fairly comprehensive dataset via Baseball Savant containing every pitch he threw in 2025 with roughly 100 features tied to every pitch.

## League-Wide Pitch Usage

```python
bs = pd.read_csv("/Users/jonlee/Desktop/Projects/Predicting Pitches/Data/pitch-arsenal-stats.csv")
bs.head()
```

```python
# Total pitches thrown league-wide
total_pitches = bs["pitches"].sum()

# League-wide pitch usage
league_pitch_usage = (
    bs.groupby(["pitch_type", "pitch_name"], as_index=False)["pitches"]
      .sum()
      .assign(
          pitch_usage=lambda x: 100 * x["pitches"] / total_pitches
      )
      .sort_values("pitch_usage", ascending=False)
)

league_pitch_usage
```

```python
# Group each pitch into a broader category
pitch_groups = {
    "FF": "Fastball",      # 4-Seam Fastball
    "SI": "Fastball",      # Sinker
    "FC": "Fastball",      # Cutter

    "SL": "Breaking Ball", # Slider
    "ST": "Breaking Ball", # Sweeper
    "CU": "Breaking Ball", # Curveball
    "KC": "Breaking Ball", # Knuckle Curve
    "SV": "Breaking Ball", # Slurve
    "CS": "Breaking Ball", # Slow Curve
    "SC": "Breaking Ball", # Screwball

    "CH": "Offspeed",      # Changeup
    "FS": "Offspeed",      # Splitter
    "FO": "Offspeed",      # Forkball
    "KN": "Offspeed"       # Knuckleball
}

league_pitch_usage["pitch_group"] = league_pitch_usage["pitch_type"].map(pitch_groups)

#Summary table
pitch_group_usage = (
    league_pitch_usage
    .groupby("pitch_group", as_index=False)["pitch_usage"]
    .sum()
    .sort_values("pitch_usage", ascending=False)
)

pitch_group_usage
```

This matches up fairly well with our previous data. Any discrepancies can be explained with various data collection criteria variance.

### Data Restructuring

First, we want to transform pitch result data to pre-pitch data for the next pitch. Each pitch contains features that describe the result (i.e. pitch velocity, strike vs. ball, etc.) The goal is to move these result variables from a given pitch to become pre-pitch variables of the next pitch. For example, if the first pitch of the game is 95 mph, and is contained within a variable called velocity_of_pitch, we seek to change this data point to become 95 mph for the second pitch within a variable called velocity_of_prev_pitch. The goal is to prevent leakage, and add more predictive strength to our model.

```python
# -------------------------
# Create KG1 safely
# -------------------------

kg1 = kg0.copy()

# Drop rows without pitch type
kg1 = kg1.dropna(subset=["pitch_type"]).copy()

# Combine sinkers into fastballs
mask = kg1["pitch_type"] == "SI"
kg1.loc[mask, ["pitch_type", "pitch_name"]] = ["FF", "4-Seam Fastball"]

# Sort chronologically instead of relying on file order
kg1 = kg1.sort_values(
    ["game_date", "game_pk", "at_bat_number", "pitch_number"]
).reset_index(drop=True)

# Explicit previous-pitch result columns
result_cols = [
    "release_speed",
    "release_pos_x",
    "release_pos_z",
    "description",
    "zone",
    "type",
    "pfx_x",
    "pfx_z",
    "plate_x",
    "plate_z",
    "vx0",
    "vy0",
    "vz0",
    "ax",
    "ay",
    "az",
    "effective_speed",
    "release_spin_rate",
    "release_extension",
    "release_pos_y",
    "spin_axis",
    "delta_home_win_exp",
    "delta_run_exp",
    "bat_speed",
    "swing_length",
    "miss_distance",
    "estimated_slg_using_speedangle",
    "delta_pitcher_run_exp",
    "hyper_speed",
    "api_break_z_with_gravity",
    "api_break_x_arm",
    "api_break_x_batter_in",
    "arm_angle",
    "attack_angle",
    "attack_direction",
    "swing_path_tilt",
    "intercept_ball_minus_batter_pos_x_inches",
    "intercept_ball_minus_batter_pos_y_inches",
    "hit_distance_sc",
    "launch_speed",
    "launch_angle"
]

# Only keep columns that actually exist
result_cols = [col for col in result_cols if col in kg1.columns]

# Shift previous-pitch variables within each game
kg1_prev = (
    kg1.groupby("game_pk")[result_cols]
       .shift(1)
       .add_suffix("_of_prev_pitch")
)

# Drop original current-pitch result columns
kg1 = kg1.drop(columns=result_cols, errors="ignore")

# Add previous-pitch versions
kg1 = pd.concat([kg1, kg1_prev], axis=1)
```

```python
# Move game_date and pitch_number to the front
front_cols = ["game_date", "pitch_number"]

kg1 = kg1[front_cols + [col for col in kg1.columns if col not in front_cols]]
```

```python
# Rename transformed variables to variable_name_of_prev_pitch
kg1 = kg1.rename(
    columns={col: f"{col}_of_prev_pitch" for col in result_cols}
)
```

```python
#fixing at_bat_number
# Fix at_bat_number so it increases by 1 for each new plate appearance within each game

kg1 = kg1.copy()

# A new plate appearance starts whenever pitch_number == 1
kg1["new_plate_appearance"] = kg1["pitch_number"].eq(1)

# Count plate appearances within each game
kg1["at_bat_number_fixed"] = (
    kg1.groupby("game_pk")["new_plate_appearance"]
       .cumsum()
)

# Replace the old at_bat_number
kg1["at_bat_number"] = kg1["at_bat_number_fixed"]

# Drop helper columns
kg1 = kg1.drop(columns=["new_plate_appearance", "at_bat_number_fixed"])
```

```python
# Columns to place at the front (in this order)
front_cols = [
    "game_date",
    "pitch_number",
    "at_bat_number",
    "inning",
    "n_thruorder_pitcher"
]

# Reorder columns
kg1 = kg1[front_cols + [col for col in kg1.columns if col not in front_cols]]
```

### Transform at-bat variables
We seek to do the same transformation with at-bat result variables: moving the variables that describe the result of a given at-bat to associate with the previous at-bat. For example, if a given at-bat results in a strikeout contained within a variable called "result_of_ab", we would move this data point to the subsequent at-bat and associate this result to the pitches of the next at-bat within a variable called "result_of_prev_ab"

```python
#Transforming at-bat result variables: the goal is to give each of the pitches of the following at-bat the associated results of the previous at-bat
ab_result_cols = [
    "events",
    "hit_location",
    "bb_type",
    "estimated_ba_using_speedangle",
    "estimated_woba_using_speedangle",
    "woba_value",
    "woba_denom",
    "babip_value",
    "iso_value",
    "launch_speed_angle"
]

# Save original row order
kg1["_row_order"] = range(len(kg1))

# Get one result row per at-bat:
# within each game + at-bat, take the last non-null value for each AB result column
ab_results = (
    kg1
    .groupby(["game_pk", "at_bat_number"], sort=False)[ab_result_cols]
    .last()
    .reset_index()
)

# Shift AB results forward so AB 3 receives AB 2's result, AB 4 receives AB 3's result, etc.
ab_results[ab_result_cols] = (
    ab_results
    .groupby("game_pk", sort=False)[ab_result_cols]
    .shift(1)
)

# Rename columns to say they describe the previous at-bat
ab_results = ab_results.rename(
    columns={col: f"{col}_of_prev_ab" for col in ab_result_cols}
)

# Merge previous-AB results back onto every pitch in the current at-bat
kg1 = kg1.merge(
    ab_results,
    on=["game_pk", "at_bat_number"],
    how="left"
)

# Restore original row order
kg1 = (
    kg1
    .sort_values("_row_order")
    .drop(columns="_row_order")
    .reset_index(drop=True)
)
```

```python
# Drop original current at-bat result columns after creating previous-at-bat versions
kg1 = kg1.drop(columns=ab_result_cols, errors="ignore")
```

```python
#Move edited features to left for checking
prev_ab_cols = [
    "events_of_prev_ab",
    "hit_location_of_prev_ab",
    "bb_type_of_prev_ab",
    "estimated_ba_using_speedangle_of_prev_ab",
    "estimated_woba_using_speedangle_of_prev_ab",
    "woba_value_of_prev_ab",
    "woba_denom_of_prev_ab",
    "babip_value_of_prev_ab",
    "iso_value_of_prev_ab",
    "launch_speed_angle_of_prev_ab"
]

front_cols = [
    "game_date",
    "pitch_number",
    "at_bat_number",
    *prev_ab_cols,
    "inning",
    "n_thruorder_pitcher"
]

kg1 = kg1[
    front_cols +
    [col for col in kg1.columns if col not in front_cols]
]
```

```python
#delete pitch_name variable as it is redundant
kg1 = kg1.drop(columns=["pitch_name"], errors="ignore")
```
